```python
import pandas as pd

df = pd.read_csv("data/transactions.csv")
```

```python
#total amount by transaction type

total_amount_by_type = (
    df.groupby("transaction_type")
    ["amount"].sum()
)

print(total_amount_by_type)
```

```python
#average amount by transaction type

transaction_average = (
    df.groupby("transaction_type")
    ["amount"]
    .agg("mean")
)

print(transaction_average)
```

```python
#transaction count by status

transaction_by_status = (
    df.groupby("status")
    .agg("count")
)

print(transaction_by_status)
```

```python
# total amount by customer
records = df[
    ["transaction_id", "customer_id", "amount"]
]

total_amount_by_cust = (
    records.groupby("customer_id")
    .agg("sum")
)

print(total_amount_by_cust)

```

```python
# maximum transaction per customer

max_transaction_by_cust = (
    records.groupby("customer_id")
    .agg("max")
)

print(max_transaction_by_cust)

```
